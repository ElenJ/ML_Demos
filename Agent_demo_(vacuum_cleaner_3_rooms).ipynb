# Agent demo in 3 rooms

### Reflex Agent Introduction

A **reflex agent** is the simplest type of agent in Artificial Intelligence. It operates by directly mapping current perceptions to actions, without any internal state or memory of past actions. Essentially, a reflex agent works on a simple "if-then" rule: if a certain condition is perceived, then a specific action is performed. They are effective in environments that are fully observable and where the correct action can be determined solely by the current perception, but they cannot handle partially observable environments or situations requiring memory and planning.

```python
# --- A super simple reflex vacuum agent ---

# Environment: three rooms, A, B, C
environment = {
    "A": "Dirty",
    "B": "Clean",
    "C": "Clean"
}

# Agent starting position
agent_position = "A"

def reflex_agent(position, status):
    # a. If the current status is 'Dirty', the action should be 'Suck'.
    if status == "Dirty":
        return "Suck"
    # b. If the agent is in position 'A' and the status is 'Clean', the action should be 'Right' (move to B).
    elif position == "A":
        return "Right"
    # c. If the agent is in position 'B' and the status is 'Clean', the action should be 'Right' (move to C).
    elif position == "B":
        return "Right"
    # d. If the agent is in position 'C' and the status is 'Clean', the action should be 'Left' (move to B).
    else: # position == "C"
        return "Left"

# A tiny simulation
# Adjust the simulation loop to run for a sufficient number of steps (e.g., 6-8 steps)
for step in range(8):
    status = environment[agent_position]
    action = reflex_agent(agent_position, status)

    print(f"Step {step+1}")
    print(f" Agent perceives: location={agent_position}, status={status}")
    print(f" Agent action: {action}")

    # environment update
    if action == "Suck":
        environment[agent_position] = "Clean"
    # Update the environment update logic to correctly change agent_position based on 'Left' or 'Right' actions for a 3-room linear topology (A <-> B <-> C).
    elif action == "Right":
        if agent_position == "A":
            agent_position = "B"
        elif agent_position == "B":
            agent_position = "C"
    elif action == "Left":
        if agent_position == "C":
            agent_position = "B"
        elif agent_position == "B":
            agent_position = "A" # This case will only be hit if agent moves Left from B.

    print(f" Environment now: {environment}")
    print()
```

### Model-Based Reflex Agent

Unlike the simple reflex agent, a model-based agent maintains an internal representation of the world (its 'model'). It uses this model, along with its current perceptions, to decide on an action. This allows it to make more informed decisions, especially when immediate perception is insufficient.

```python
# Environment: three rooms, A, B, C
environment_mb = {
    "A": "Dirty",
    "B": "Clean",
    "C": "Clean"
}

# Agent starting position
agent_position_mb = "A"

# Agent's internal model of the environment
agent_model = {
    "A": "Unknown",
    "B": "Unknown",
    "C": "Unknown"
}

def model_based_agent(position, percept_status, model):
    # Update the agent's internal model based on current perception
    model[position] = percept_status

    if percept_status == "Dirty":
        return "Suck", model
    elif position == "A":
        # If A is clean, and B is dirty or unknown, move right to B
        if model["B"] == "Dirty" or model["B"] == "Unknown":
            return "Right", model
        else:
            # Otherwise, just move right to B (assuming a sweep pattern)
            return "Right", model
    elif position == "B":
        # If A is dirty, go back to A
        if model["A"] == "Dirty":
            return "Left", model
        # If A is clean, but C is dirty or unknown, move right to C
        elif model["C"] == "Dirty" or model["C"] == "Unknown":
            return "Right", model
        else: # Both A and C are known clean, or unknown but not dirty
            # If A and C are known clean, move left to A (sweep back)
            return "Left", model
    else: # position == "C"
        # If C is clean, and B is dirty or unknown, move left to B
        if model["B"] == "Dirty" or model["B"] == "Unknown":
            return "Left", model
        else:
            # Otherwise, just move left to B (assuming a sweep pattern)
            return "Left", model

# A tiny simulation for the model-based agent
print("\n--- Model-Based Agent Simulation (3 Rooms) ---")
for step in range(10): # Run for more steps to observe movements across all three rooms
    status = environment_mb[agent_position_mb]
    action, agent_model = model_based_agent(agent_position_mb, status, agent_model)

    print(f"Step {step+1}")
    print(f" Agent perceives: location={agent_position_mb}, status={status}")
    print(f" Agent's internal model: {agent_model}") # Show internal model
    print(f" Agent action: {action}")

    # environment update
    if action == "Suck":
        environment_mb[agent_position_mb] = "Clean"
    elif action == "Right":
        if agent_position_mb == "A":
            agent_position_mb = "B"
        elif agent_position_mb == "B":
            agent_position_mb = "C"
    elif action == "Left":
        if agent_position_mb == "C":
            agent_position_mb = "B"
        elif agent_position_mb == "B":
            agent_position_mb = "A"

    print(f" Environment now: {environment_mb}")
    print()
```

### Model-Based Agent with Probabilistic Dirt Generation

To make the environment more realistic and challenging, we can introduce a probability that a clean room might become dirty again after some time. This mimics real-world scenarios where cleanliness is not permanent. The model-based agent will need to adapt to these dynamic changes.

```python
import random

# Environment: three rooms, A, B, C
environment_prob = {
    "A": "Clean", # Starting clean to see dirt generation
    "B": "Clean",
    "C": "Clean"
}

# Agent starting position
agent_position_prob = "A"

# Agent's internal model of the environment
agent_model_prob = {
    "A": "Unknown",
    "B": "Unknown",
    "C": "Unknown"
}

# Probability of a clean room becoming dirty in a given step
probability_of_dirt_generation = 0.3 # 30% chance

def model_based_agent_prob(position, percept_status, model):
    # Update the agent's internal model based on current perception
    model[position] = percept_status

    if percept_status == "Dirty":
        return "Suck", model
    elif position == "A":
        # If A is clean, and B is dirty or unknown, move right to B
        if model["B"] == "Dirty" or model["B"] == "Unknown":
            return "Right", model
        else:
            # Otherwise, just move right to B (assuming a sweep pattern)
            return "Right", model
    elif position == "B":
        # If A is dirty, go back to A
        if model["A"] == "Dirty":
            return "Left", model
        # If A is clean, but C is dirty or unknown, move right to C
        elif model["C"] == "Dirty" or model["C"] == "Unknown":
            return "Right", model
        else: # Both A and C are known clean, or unknown but not dirty
            # If A and C are known clean, move left to A (sweep back)
            return "Left", model
    else: # position == "C"
        # If C is clean, and B is dirty or unknown, move left to B
        if model["B"] == "Dirty" or model["B"] == "Unknown":
            return "Left", model
        else:
            # Otherwise, just move left to B (assuming a sweep pattern)
            return "Left", model

# A tiny simulation for the model-based agent with probabilistic dirt generation
print("\n--- Model-Based Agent with Probabilistic Dirt Generation Simulation (3 Rooms) ---")
for step in range(15): # Run for more steps to observe dirt generation
    print(f"\nStep {step+1}")

    # --- Probabilistic dirt generation --- before agent perceives
    for room in environment_prob:
        if environment_prob[room] == "Clean" and random.random() < probability_of_dirt_generation:
            environment_prob[room] = "Dirty"
            print(f"  Room {room} randomly became Dirty.")

    status = environment_prob[agent_position_prob]
    action, agent_model_prob = model_based_agent_prob(agent_position_prob, status, agent_model_prob)

    print(f" Agent perceives: location={agent_position_prob}, status={status}")
    print(f" Agent's internal model: {agent_model_prob}") # Show internal model
    print(f" Agent action: {action}")

    # environment update based on agent's action
    if action == "Suck":
        environment_prob[agent_position_prob] = "Clean"
    elif action == "Right":
        if agent_position_prob == "A":
            agent_position_prob = "B"
        elif agent_position_prob == "B":
            agent_position_prob = "C"
    elif action == "Left":
        if agent_position_prob == "C":
            agent_position_prob = "B"
        elif agent_position_prob == "B":
            agent_position_prob = "A"

    print(f" Environment now: {environment_prob}")
```

### Goal-Based vs. Utility-Based Agents

Both goal-based and utility-based agents are types of model-based agents, meaning they maintain an internal representation of the world. The key difference lies in *how* they make decisions.

*   **Goal-Based Agent:** This agent has a specific target state or set of states it wants to achieve. Its actions are chosen to reach that goal. Once the goal is achieved, the agent typically stops or enters a 'done' state, or it might look for a new goal. It doesn't necessarily consider the *quality* or *efficiency* of the path to the goal, just that the goal is met.
    *   **Example (Vacuum Agent):** The goal might simply be: "All rooms are clean." Once its model confirms both rooms are clean, it might stop moving or patrol minimally, even if dirt reappears later.

*   **Utility-Based Agent:** This agent aims to maximize its 'utility' (a quantitative measure of its performance or happiness) over time. It considers not only reaching a goal but also *how well* it reaches it, often by evaluating the consequences of its actions in terms of expected outcomes and their associated utility. It will choose actions that maximize the expected utility of the sequence of actions.
    *   **Example (Vacuum Agent):** The agent might have a utility function that assigns points for clean rooms, deducts points for dirty rooms, and deducts points for actions (like moving or sucking). It would constantly strive to maximize this score, perhaps by actively patrolling to *prevent* dirt accumulation, or by choosing the most efficient path to clean rooms, rather than just waiting for them to become dirty.

### Demonstration: Goal-Based Agent

We'll adapt our previous `model_based_agent_prob` to explicitly act as a goal-based agent. Its goal will be to get both rooms clean. Once its internal model indicates both rooms are clean, it will simply 'idle' (do nothing) until it perceives dirt again.

```python
import random

environment_goal = {
    "A": "Dirty",
    "B": "Clean",
    "C": "Clean"
}
agent_position_goal = "A"
agent_model_goal = {
    "A": "Unknown",
    "B": "Unknown",
    "C": "Unknown"
}
probability_of_dirt_generation_goal = 0.3

def goal_based_agent(position, percept_status, model):
    # Update the agent's internal model based on current perception
    model[position] = percept_status

    # Define the goal: all rooms are clean according to its model
    goal_achieved = (model["A"] == "Clean" and model["B"] == "Clean" and model["C"] == "Clean")

    if percept_status == "Dirty":
        return "Suck", model
    elif goal_achieved: # If goal is achieved, the agent idles
        return "Idle", model # A new action to represent no meaningful movement
    elif position == "A":
        # If A is clean, and B is dirty or unknown, move right to B
        if model["B"] == "Dirty" or model["B"] == "Unknown":
            return "Right", model
        else:
            # Otherwise, just move right to B (assuming a sweep pattern)
            return "Right", model
    elif position == "B":
        # If A is dirty, go back to A
        if model["A"] == "Dirty":
            return "Left", model
        # If A is clean, but C is dirty or unknown, move right to C
        elif model["C"] == "Dirty" or model["C"] == "Unknown":
            return "Right", model
        else: # Both A and C are known clean, or unknown but not dirty
            # If A and C are known clean, move left to A (sweep back)
            return "Left", model
    else: # position == "C"
        # If C is clean, and B is dirty or unknown, move left to B
        if model["B"] == "Dirty" or model["B"] == "Unknown":
            return "Left", model
        else:
            # Otherwise, just move left to B (assuming a sweep pattern)
            return "Left", model

print("\n--- Goal-Based Agent Simulation (3 Rooms) ---")
for step in range(15): # Run for more steps
    print(f"\nStep {step+1}")

    # --- Probabilistic dirt generation ---
    for room in environment_goal:
        if environment_goal[room] == "Clean" and random.random() < probability_of_dirt_generation_goal:
            environment_goal[room] = "Dirty"
            print(f"  Room {room} randomly became Dirty.")

    status = environment_goal[agent_position_goal]
    action, agent_model_goal = goal_based_agent(agent_position_goal, status, agent_model_goal)

    print(f" Agent perceives: location={agent_position_goal}, status={status}")
    print(f" Agent's internal model: {agent_model_goal}")
    print(f" Agent action: {action}")

    # environment update
    if action == "Suck":
        environment_goal[agent_position_goal] = "Clean"
    elif action == "Right":
        if agent_position_goal == "A":
            agent_position_goal = "B"
        elif agent_position_goal == "B":
            agent_position_goal = "C"
    elif action == "Left":
        if agent_position_goal == "C":
            agent_position_goal = "B"
        elif agent_position_goal == "B":
            agent_position_goal = "A"
    # 'Idle' action implies no change to position or environment state

    print(f" Environment now: {environment_goal}")
```

### Demonstration: Utility-Based Agent

For a utility-based agent, we'll introduce a simple utility function. This agent will aim to keep the environment as clean as possible while also being mindful of unnecessary moves. Its decisions will try to maximize its perceived utility. In this simplified version, if the current room is clean, it will proactively check the other room, assuming a higher utility in maintaining overall cleanliness rather than just reacting to dirt in its current location.

```python
import random

environment_utility = {
    "A": "Dirty",
    "B": "Clean",
    "C": "Clean"
}
agent_position_utility = "A"
agent_model_utility = {
    "A": "Unknown",
    "B": "Unknown",
    "C": "Unknown"
}
probability_of_dirt_generation_utility = 0.3

# A simple utility score tracker (for observation, not directly for decision making in this simple model)
# A real utility agent would use this in its decision process (e.g., lookahead search)
agent_utility_score = 0

def utility_based_agent(position, percept_status, model):
    global agent_utility_score # To modify the global score

    # Update model
    model[position] = percept_status

    action = ""
    if percept_status == "Dirty":
        action = "Suck"
        agent_utility_score += 10 # Reward for cleaning
    elif position == "A":
        # If A is clean, and B is dirty or unknown, move right to B
        if model["B"] == "Dirty" or model["B"] == "Unknown":
            action = "Right"
            agent_utility_score -= 1 # Penalty for moving
        else:
            # Otherwise, just move right to B (assuming a sweep pattern)
            action = "Right"
            agent_utility_score -= 1 # Penalty for moving
    elif position == "B":
        # If A is dirty, go back to A
        if model["A"] == "Dirty":
            action = "Left"
            agent_utility_score -= 1 # Penalty for moving
        # If A is clean, but C is dirty or unknown, move right to C
        elif model["C"] == "Dirty" or model["C"] == "Unknown":
            action = "Right"
            agent_utility_score -= 1 # Penalty for moving
        else: # Both A and C are known clean, or unknown but not dirty
            # If A and C are known clean, move left to A (sweep back)
            action = "Left"
            agent_utility_score -= 1 # Penalty for moving
    else: # position == "C"
        # If C is clean, and B is dirty or unknown, move left to B
        if model["B"] == "Dirty" or model["B"] == "Unknown":
            action = "Left"
            agent_utility_score -= 1 # Penalty for moving
        else:
            # Otherwise, just move left to B (assuming a sweep pattern)
            action = "Left"
            agent_utility_score -= 1 # Penalty for moving

    # Consider a penalty if a room becomes dirty while agent is elsewhere
    # (This is more for tracking utility, the agent's decision logic above is simplified)
    if (model['A'] == 'Dirty' or model['B'] == 'Dirty' or model['C'] == 'Dirty'):
        agent_utility_score -= 5 # Penalty for allowing dirt to persist in other room

    return action, model

print("\n--- Utility-Based Agent Simulation (3 Rooms) ---")
for step in range(15): # Run for more steps
    print(f"\nStep {step+1}")

    # --- Probabilistic dirt generation ---
    for room in environment_utility:
        if environment_utility[room] == "Clean" and random.random() < probability_of_dirt_generation_utility:
            environment_utility[room] = "Dirty"
            print(f"  Room {room} randomly became Dirty.")

    status = environment_utility[agent_position_utility]
    action, agent_model_utility = utility_based_agent(agent_position_utility, status, agent_model_utility)

    print(f" Agent perceives: location={agent_position_utility}, status={status}")
    print(f" Agent's internal model: {agent_model_utility}")
    print(f" Agent action: {action}")
    print(f" Current Utility Score: {agent_utility_score}") # Show utility score

    # environment update
    if action == "Suck":
        environment_utility[agent_position_utility] = "Clean"
    elif action == "Right":
        if agent_position_utility == "A":
            agent_position_utility = "B"
        elif agent_position_utility == "B":
            agent_position_utility = "C"
    elif action == "Left":
        if agent_position_utility == "C":
            agent_position_utility = "B"
        elif agent_position_utility == "B":
            agent_position_utility = "A"

    print(f" Environment now: {environment_utility}")
```
